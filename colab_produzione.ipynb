## 1. Setup: codice e dipendenze

```python
# ============================================================
#  DA DOVE SCARICARE IL CODICE
#  Se cambi repository, cambia SOLO la riga qui sotto.
# ============================================================
REPO_URL = "https://github.com/Giacomod2001/mmm-bayesiano-recruiting.git"
# ============================================================

import os, sys, subprocess

CARTELLA = REPO_URL.rstrip("/").split("/")[-1].replace(".git", "")

if os.path.isdir(CARTELLA):
    print("Il codice c'e' gia' (cartella", CARTELLA + "). Non lo riscarico.")
else:
    print("Scarico il codice da:", REPO_URL)
    esito = subprocess.run(["git", "clone", "--depth", "1", REPO_URL],
                           capture_output=True, text=True)
    if esito.returncode != 0:
        print()
        print("!" * 72)
        print("!!!   CLONE FALLITO: il codice NON e' stato scaricato.")
        print("!" * 72)
        print()
        print("Causa piu' probabile: il repository e' PRIVATO.")
        print("Colab non ha le credenziali, quindi non riesce a scaricarlo.")
        print()
        print("Cosa fare (una delle due):")
        print("  1) scrivi in REPO_URL, qui sopra, l'indirizzo del repo PUBBLICO")
        print("     e riesegui questa cella;")
        print("  2) oppure carica i file a mano dal pannello a sinistra di Colab.")
        print()
        print("Messaggio di git:")
        print("   " + (esito.stderr.strip() or "(nessun messaggio)"))
        print()
        raise SystemExit("Mi fermo qui: senza il codice le celle successive non possono funzionare.")

os.chdir(CARTELLA)
print("Codice pronto in:", os.getcwd())

print()
print("Installo le librerie (di solito 1-2 minuti)...")
esito = subprocess.run([sys.executable, "-m", "pip", "install", "-q",
                        "-r", "pipeline/requirements.txt"])
if esito.returncode != 0:
    print("Installazione non riuscita. Riesegui questa cella una seconda volta.")
    raise SystemExit("Mi fermo qui: mancano le librerie.")

print("Fatto. Puoi passare alla cella successiva.")
```

## 2. Carica i 4 CSV puliti

```python
import os
from google.colab import files

CANON = 'pipeline/data/canonical'
os.makedirs(CANON, exist_ok=True)
print('Seleziona: media.csv, demand.csv, seasonality.csv, outcome.csv')
caricati = files.upload()
for nome in caricati:
    os.replace(nome, os.path.join(CANON, nome))
presenti = sorted(os.listdir(CANON))
print('File in canonical:', presenti)
attesi = {'media.csv', 'demand.csv', 'seasonality.csv', 'outcome.csv'}
mancanti = attesi - set(presenti)
assert not mancanti, f'Mancano questi file: {mancanti}'
print('OK: tutti e 4 i file canonici sono pronti.')
```

### (opzionale) Carica il cruscotto come Foglio Google

Crea `risultati.xlsx` come **Foglio Google** nel tuo Drive e stampa il link (usa il login Google di Colab: ti chiede l'autorizzazione).

```python
from google.colab import auth
auth.authenticate_user()
from googleapiclient.discovery import build
from googleapiclient.http import MediaFileUpload

drive = build('drive', 'v3')
meta = {'name': 'MMM Risultati', 'mimeType': 'application/vnd.google-apps.spreadsheet'}
media = MediaFileUpload('pipeline/data/output/risultati.xlsx',
                        mimetype='application/vnd.openxmlformats-officedocument.spreadsheetml.sheet')
f = drive.files().create(body=meta, media_body=media, fields='id,webViewLink').execute()
print('Foglio Google creato nel tuo Drive:')
print(f['webViewLink'])
```
